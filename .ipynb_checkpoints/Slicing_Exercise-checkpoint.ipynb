```python
import numpy as np
arr=np.arange(25).reshape(5,5)
print(arr)
print(arr[1:3,1:3])
```

<p style="font-family:Monotype"><b>Q1</b>. Extract </p> 
<pre>
   [[ 6  7]
    [11 12]]
</pre>

```python
result=arr[1:3,1:3]
print(result)
```

<p style="font-family:Monotype"><b>Q2.</b> Extract the last two rows </p>

```python
result=arr[-2:]
print(result)
```

<p style="font-family:Monotype"><b>Q3.</b> Extract every alternate row</p>

```python
result=arr[::2]
print(result)
```

<p style="font-family:Monotype"><b>Q4.</b> Extract every alternate column</p>


```python
result=arr[:,::2]
print(result)
```

<p style="font-family:Monotype"><b>Q5.</b> Reverse every row </p>


```python
result=arr[:,::-1]
print(result)
```

<p style="font-family:Monotype"><b>Q6.</b> Reverse every column </p>


```python
result=arr[::-1,:]
print(result)
```

<p style="font-family:Monotype"><b>Q7.</b> Extract the center 3x3 matrix </p>


```python
result=arr[1:4,1:4]
print(result)
```

<p style="font-family:Monotype"><b>Q8.</b> Extract the border elements only with out using loops </p>

```python
result=arr.copy()
result[1:4,1:4]=0
print(result)
```
