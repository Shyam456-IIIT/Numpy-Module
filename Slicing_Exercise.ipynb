```python
import numpy as np
arr=np.arange(25).reshape(5,5)
print(arr)
print(arr[1:3,1:3])
```

<p style="font-family:Monotype"><b>Q1</b>. Extract </p>

```python
result=arr[1:3,1:3]
print(result)
```

<p style="font-family:Monotype"><b>Q2.</b> Extract the last two rows </p>

```python
result=arr[-2:]
print(result)
```

<p style="font-family:Monotype"><b>Q3.</b> Extract every alternate row</p>

```python
result=arr[::2]
print(result)
```

<p style="font-family:Monotype"><b>Q4.</b> Extract every alternate column</p>


```python
result=arr[:,::2]
print(result)
```

<p style="font-family:Monotype"><b>Q5.</b> Reverse every row </p>


```python
result=arr[:,::-1]
print(result)
```

<p style="font-family:Monotype"><b>Q6.</b> Reverse every column </p>


```python
result=arr[::-1,:]
print(result)
```

<p style="font-family:Monotype"><b>Q7.</b> Extract the center 3x3 matrix </p>


```python
result=arr[1:4,1:4]
print(result)
```

<p style="font-family:Monotype"><b>Q8.</b> Extract the border elements only with out using loops </p>

```python
result=arr.copy()
result[1:4,1:4]=0
mask = np.zeros_like(result,dtype=bool)
mask[1:4,1:4]=True
masked=np.ma.masked_array(result,mask)
print(masked)
```

<p style="font-family: Cascadia Code"> 
  <code style="background: lightgrey">np.ma.masked_array</code> 
  is a NumPy function that creates a masked array by combining the data with a boolean mask, 
  allowing selected elements to be hidden without modifying the original data. 
  For example, you can use it to hide missing or invalid data points during calculations.
</p>

<p style="font-family:Monotype"><b>Q12.</b> Swap the first and last rows </p>

```python
result=arr.copy()
result[0,:]=arr[-1,:]
result[-1,:]=arr[0,:]
print(result)
```
