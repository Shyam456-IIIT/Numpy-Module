The UCI Red Wine Quality Dataset is one of the most popular datasets for beginners learning data manipulation and machine learning. Created by researchers in Portugal, it contains chemical analyses of 1,599 variants of the Portuguese "Vinho Verde" red wine.

```python
import numpy as np

# Direct link to the dataset
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv'

# Load the data (skipping the first row which contains text headers, and splitting by semicolon)
wine_data = np.genfromtxt(url, delimiter=';', skip_header=1)
```

```python
import pandas as pd
from io import StringIO
```

```python
data="""
Index,Feature,Description
0,Fixed Acidity,Acids involved with wine that don't evaporate easily
1,Volatile Acidity,Acetic acid (too much leads to a vinegar taste)
2,Citric Acid,Adds 'freshness' and flavor to wines
3,Residual Sugar,Amount of sugar remaining after fermentation
4,Chlorides,Amount of salt in the wine
5,Free Sulfur Dioxide,Prevents microbial growth and oxidation
6,Total Sulfur Dioxide,Amount of free and bound forms of SO2
7,Density,Density of the wine (usually close to water)
8,pH,Acidity scale (typically 3.0 to 4.0)
9,Sulphates,An additive that contributes to SO2 levels
10,Alcohol,Percentage of alcohol content
11,Quality (Target),A score from 0 to 10 given by human tasters 
"""
```

##### **Challenge 1:** The Alcohol-Quality Correlation (Intermediate)Does higher alcohol content mean better wine?
##### Calculate and print the mean alcohol content (index 10) for "good" wines (quality $\ge 7$).
##### Calculate and print the mean alcohol content for "bad" wines (quality $\le 4$).

```python
# Here, we are checking the quality of the wine
# Extracting the best quality wine.
good_wines=wine_data[:,11]>=7
bad_wines=wine_data[:,11]<=4

# Mean value of those quality wines 
mean_good_wines=wine_data[:,10][good_wines].mean()
" mean_good_wines=wine_data[good_wines,10].mean()"
print(mean_good_wines)

# Mean value of those not quality wines
mean_bad_wines=wine_data[:,10][bad_wines].mean()
" mean_bad_wines=wine_data[bad_wines,10].mean()"
print(mean_bad_wines)
```
