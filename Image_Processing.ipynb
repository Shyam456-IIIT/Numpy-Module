```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Image Reading
img=plt.imread("nature.jpeg")
plt.imshow(img)
plt.axis('on')
plt.show()
```

```python
# Image shape, dtype
img_matrix=np.array(img)
print(img_matrix.shape)
print(img_matrix.dtype)
```

```python
def crop_image(img, start_y, end_y, start_x, end_x):
    cropped_img=img_matrix[start_y:end_y,start_x:end_x]
    return cropped_img
```

```python
cropped_img1=crop_image(img,75,250,0,300)
plt.imshow(cropped_img1)
plt.axis("off")
plt.show()
```

```python
cropped_img2=crop_image(img,75,240,200,400)
plt.imshow(cropped_img2)
plt.axis("off")
plt.show()
```

```python
def adjust_brightness(img, factor):
    img_int=img.astype(int)

    # add brightness factor
    brightened = img_int + factor 

    brightened_img = np.clip(brightened,0,255)
    return brightened_img.astype(np.uint8)
```

```python
darken_img=adjust_brightness(img, -70)
plt.imshow(darken_img)
plt.axis("off")
plt.imsave("Darken_img.jpg",darken_img)
plt.show()
print("Image saved successfully!")
```

```python
bright_img=adjust_brightness(img, 70)
plt.imshow(bright_img)
plt.axis("off")
plt.imsave("brightened_img.jpg",bright_img)
plt.show()
print("Image saved successfully!")
```
